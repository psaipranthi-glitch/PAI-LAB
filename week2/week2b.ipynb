```python
import heapq

def water_jug_astar(capX, capY, target):
    
    heuristic = lambda x, y: abs(x - target) + abs(y - target)

    
    pq = [(heuristic(0,0), 0, 0, 0, [(0,0)])]
    visited = set()

    while pq:
        f, g, x, y, path = heapq.heappop(pq)

        
        if (x, y) in visited:
            continue
        visited.add((x, y))

        
        if x == target:
            return path

    
        pour_X_to_Y = min(x, capY - y)
        pour_Y_to_X = min(y, capX - x)

        
        moves = [
            (capX, y),               
            (x, capY),              
            (0, y),                  
            (x, 0),                  
            (x - pour_X_to_Y, y + pour_X_to_Y), 
            (x + pour_Y_to_X, y - pour_Y_to_X)   
        ]

        
        for nx, ny in moves:
            if (nx, ny) not in visited:
                g_new = g + 1
                f_new = g_new + heuristic(nx, ny)
                heapq.heappush(pq, (f_new, g_new, nx, ny, path + [(nx, ny)]))

    return None


# Example usage
solution = water_jug_astar(7, 4, 6)

if solution:
    print("Steps to reach 6L in X:")
    for step in solution:
        print(step)
else:
    print("No solution found.")
```
