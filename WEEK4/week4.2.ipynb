```python
proctors = {
    "P1": {"skills": ["Math", "AI"], "times": ["T1", "T2"]},
    "P2": {"skills": ["Physics", "Math"], "times": ["T2", "T3"]},
    "P3": {"skills": ["AI", "CS"], "times": ["T1", "T3"]}
}

exams = {
    "E1": {"skill": "Math", "time": "T1"},
    "E2": {"skill": "AI", "time": "T2"},
    "E3": {"skill": "CS", "time": "T3"}
}

assignment = {}

def is_valid(proctor, exam):
    if exams[exam]["skill"] not in proctors[proctor]["skills"]:
        return False
    if exams[exam]["time"] not in proctors[proctor]["times"]:
        return False
    for e, p in assignment.items():
        if p == proctor and exams[e]["time"] == exams[exam]["time"]:
            return False
    return True

def backtrack(exam_list):
    if len(assignment) == len(exam_list):
        return True
    exam = exam_list[len(assignment)]
    for proctor in proctors:
        if is_valid(proctor, exam):
            assignment[exam] = proctor
            if backtrack(exam_list):
                return True
            del assignment[exam]
    return False

exam_list = list(exams.keys())

if backtrack(exam_list):
    print("Assignments:")
    for exam, proctor in assignment.items():
        print(exam, "->", proctor)
else:
    print("No valid assignment found")
```
