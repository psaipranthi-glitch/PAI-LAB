```python
import heapq

GRID = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]

START, DEST = (0,0), (2,2)
ROWS, COLS = 3,3

h = lambda p: abs(p[0]-DEST[0])+abs(p[1]-DEST[1])
nbs = lambda p: [(x,y) for dx,dy in [(-1,0),(1,0),(0,-1),(0,1)]
                 if 0<= (x:=p[0]+dx) < ROWS and 0<= (y:=p[1]+dy) < COLS and GRID[x][y]==0]

def a_star(start):
    heap, came, g = [(h(start),0,start)], {}, {start:0}
    while heap:
        _, g_curr, cur = heapq.heappop(heap)
        if cur==DEST:
            path=[cur]
            while cur in came: cur=came[cur]; path.append(cur)
            return path[::-1]
        for nb in nbs(cur):
            t = g_curr+1
            if nb not in g or t<g[nb]: came[nb]=cur; g[nb]=t; heapq.heappush(heap,(t+h(nb),t,nb))
    return None

def print_step(path, step):
    disp=[[" "]*COLS for _ in range(ROWS)]
    for i,(x,y) in enumerate(path[:step+1]):
        if i==0: disp[x][y]="S"
        elif i==len(path[:step+1])-1: disp[x][y]="D"
        else: disp[x][y]="P"
    for i in range(ROWS):
        print("+---"*COLS+"+")
        for j in range(COLS):
            val = disp[i][j] if disp[i][j]!=" " else "#" if GRID[i][j]==1 else " "
            print(f"| {val}",end="")
        print("|")
    print("+---"*COLS+"+\n")

path = a_star(START)
if path:
    print(f"Order pick path found! Total steps: {len(path)}\n")
    for s in range(len(path)):
        print(f"Step {s}:")
        print_step(path,s)
else: print("No path found.")
```
