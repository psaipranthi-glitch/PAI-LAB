```python


from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination


model = DiscreteBayesianNetwork([
    ('Income', 'Default'),
    ('Credit', 'Default')
])

cpd_income = TabularCPD('Income', 2, [[0.6], [0.4]])
cpd_credit = TabularCPD('Credit', 2, [[0.7], [0.3]])

cpd_default = TabularCPD(
    'Default', 2,
    [[0.9, 0.6, 0.7, 0.1],
     [0.1, 0.4, 0.3, 0.9]],
    evidence=['Income', 'Credit'],
    evidence_card=[2, 2]
)


model.add_cpds(cpd_income, cpd_credit, cpd_default)


print("Model valid:", model.check_model())


inference = VariableElimination(model)


result = inference.query(variables=['Default'], evidence={'Income': 1, 'Credit': 0})
print(result)
```
