```python
import random

attacker_actions = ["Phishing", "Malware", "DDoS", "Exploit"]
defender_actions = ["Firewall", "Patch", "Monitor", "Mitigate"]

system_security = 100

def apply_actions(attacker_move, defender_move, security_score):
    impact = {
        "Phishing": 20,
        "Malware": 25,
        "DDoS": 15,
        "Exploit": 30
    }
    defense_effect = {
        "Firewall": ["DDoS", "Exploit"],
        "Patch": ["Malware", "Exploit"],
        "Monitor": ["Phishing", "Malware"],
        "Mitigate": ["DDoS", "Phishing"]
    }
    damage = impact[attacker_move]
    if attacker_move in defense_effect[defender_move]:
        damage *= 0.5
    security_score -= damage
    if security_score < 0:
        security_score = 0
    return security_score

rounds = 5
for r in range(1, rounds + 1):
    attacker_move = random.choice(attacker_actions)
    defender_move = random.choice(defender_actions)
    system_security = apply_actions(attacker_move, defender_move, system_security)
    print(f"Round {r}: Attacker -> {attacker_move}, Defender -> {defender_move}, Security -> {system_security}")

if system_security > 50:
    print("System is secure.")
else:
    print("System compromised!")
```
